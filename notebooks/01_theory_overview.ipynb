## Key Results

### 1. The Meixner Measure

The measure $\eta_c^r$ is defined by its density:

$$f_c^r(x) = \frac{1}{2\pi} \frac{\sqrt{4-x^2}}{1 + (e^c-1)(1+rx^2)}$$

This measure is crucial for understanding the transition to the Haar state.

### 2. The Cutoff Phenomenon

For $r > 0$, the total variation distance from the Haar state exhibits a **sharp cutoff**, whereas for $r = 0$, the convergence is smooth.
